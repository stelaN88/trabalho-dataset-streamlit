```python
# Criando uma copia do dataset original
df_tratado = df.copy()

# Conferindo a quantidade de registros
df_tratado.shape
```

```python
# Removendo registros completamente duplicados
df_tratado = df_tratado.drop_duplicates()

# Verificando a nova quantidade de registros
df_tratado.shape
```

```python
# Corrigindo os tipos de dados das colunas

# Identificadores devem ser armazenados como texto
df_tratado["InvoiceNo"] = df_tratado["InvoiceNo"].astype("string")
df_tratado["StockCode"] = df_tratado["StockCode"].astype("string")

# CustomerID sera inteiro, permitindo valores ausentes
df_tratado["CustomerID"] = df_tratado["CustomerID"].astype("Int64")

# Convertendo colunas de texto
df_tratado["Description"] = df_tratado["Description"].astype("string")
df_tratado["Country"] = df_tratado["Country"].astype("string")

# Conferindo os tipos corrigidos
df_tratado.info()
```

```python
# Criando uma classificacao inicial dos registros

df_tratado["TipoRegistro"] = "Registro comum"

# Identificando cancelamentos pelo numero do pedido
df_tratado.loc[
    df_tratado["InvoiceNo"].str.startswith("C", na=False),
    "TipoRegistro"
] = "Cancelamento"

# Identificando ajustes de dividas
df_tratado.loc[
    df_tratado["Description"].str.contains(
        "bad debt",
        case=False,
        na=False
    ),
    "TipoRegistro"
] = "Ajuste de divida"

# Identificando outras quantidades negativas
df_tratado.loc[
    (df_tratado["Quantity"] < 0) &
    (df_tratado["TipoRegistro"] == "Registro comum"),
    "TipoRegistro"
] = "Quantidade negativa - investigar"

# Contando os registros de cada categoria
df_tratado["TipoRegistro"].value_counts()
```

```python
# Calculando o valor total de cada registro

df_tratado["TotalPrice"] = (
    df_tratado["Quantity"] * df_tratado["UnitPrice"]
)

# Visualizando os resultados
df_tratado[
    ["InvoiceNo", "Description", "Quantity", "UnitPrice", "TotalPrice"]
].head(10)
```

```python
# Conferindo a quantidade de registros
print("Dimensoes:", df_tratado.shape)

# Conferindo valores ausentes
print("\nValores ausentes:")
print(df_tratado.isnull().sum())

# Conferindo duplicidades restantes
print("\nDuplicidades restantes:")
print(df_tratado.duplicated().sum())

# Conferindo os tipos de dados
print("\nTipos de dados:")
print(df_tratado.dtypes)

# Conferindo a classificacao dos registros
print("\nTipos de registros:")
print(df_tratado["TipoRegistro"].value_counts())
```


## Tratamento e preparação dos dados

Após a exploração inicial, foi criada uma cópia do
dataset original para realizar as modificações.

Foram aplicadas as seguintes operações:

1. Remoção de registros completamente duplicados.
2. Correção dos tipos de dados dos identificadores.
3. Preservação dos valores ausentes, evitando a exclusão
   de transações que ainda podem ser utilizadas.
4. Classificação de cancelamentos e ajustes financeiros.
5. Criação da coluna TotalPrice, calculada pela
   multiplicação da quantidade pelo preço unitário.

Os registros com quantidades negativas e preços zerados
foram preservados para permitir análises específicas.

O dataset original foi mantido sem alterações.


```python
# Conferindo a estrutura do DataFrame original
print("DataFrame original:")
print(df.shape)
print(df.columns.tolist())

# Conferindo a estrutura do DataFrame tratado
print("\nDataFrame tratado:")
print(df_tratado.shape)
print(df_tratado.columns.tolist())
```

```python
# Criando uma base contendo apenas vendas comerciais

df_vendas = df_tratado[
    (df_tratado["TipoRegistro"] == "Registro comum") &
    (df_tratado["Quantity"] > 0) &
    (df_tratado["UnitPrice"] > 0)
].copy()

# Conferindo a quantidade de registros
df_vendas.shape
```

```python
# Verificando se existem quantidades ou preços inválidos

print("Quantidades negativas:")
print((df_vendas["Quantity"] < 0).sum())

print("\nPreços negativos:")
print((df_vendas["UnitPrice"] < 0).sum())

print("\nPreços zerados:")
print((df_vendas["UnitPrice"] == 0).sum())

print("\nValores ausentes:")
print(df_vendas.isnull().sum())
```

```python
# Calculando o faturamento total da base de vendas

faturamento_total = df_vendas["TotalPrice"].sum()

print("Faturamento total: £", round(faturamento_total, 2))
```

```python
df_vendas[
    df_vendas["Quantity"] > limite_superior
].sort_values(
    by="Quantity",
    ascending=False
).head(15)
```

### Análise de valores atípicos (outliers)

Foi aplicado o método do intervalo interquartil (IQR) para identificar possíveis valores atípicos na coluna Quantity.

O limite superior calculado foi de 26 unidades. Foram identificados 27.111 registros acima desse limite, representando aproximadamente 5,16% da base de vendas.

Após analisar os resultados, decidiu-se preservar esses registros, pois quantidades elevadas não representam necessariamente erros. Elas podem corresponder a compras legítimas em grandes volumes.

Dessa forma, os possíveis outliers serão mantidos nas análises, sem exclusão automática.

```python
faturamento_mensal = (
    df_vendas.assign(
        Mes=df_vendas["InvoiceDate"].dt.to_period("M")
    )
    .groupby("Mes")["TotalPrice"]
    .sum()
)

faturamento_mensal
```

```python
import plotly.express as px

px.defaults.template = "plotly_white"

# Pergunta: como o faturamento varia entre os meses?
dados_faturamento = faturamento_mensal.reset_index()
dados_faturamento["Mes"] = dados_faturamento["Mes"].astype(str)

fig = px.bar(
    dados_faturamento,
    x="Mes",
    y="TotalPrice",
    title="Faturamento mensal",
    labels={"Mes": "Mês", "TotalPrice": "Faturamento (£)"}
)
fig.update_xaxes(tickangle=45)
fig.show()
```

### Análise da quantidade de pedidos por país

Foi realizada uma contagem das notas fiscais distintas por país, utilizando o método nunique() da biblioteca Pandas.

O Reino Unido apresentou 18.018 pedidos, ocupando a primeira posição. Em seguida aparecem Alemanha, com 457 pedidos, e França, com 392 pedidos.

Os resultados demonstram uma concentração expressiva das operações no Reino Unido.

Essa análise complementa a contagem de registros de produtos, permitindo diferenciar a quantidade de itens registrados do número de pedidos realizados.

```python
total_pedidos = df_vendas["InvoiceNo"].nunique()

pedidos_reino_unido = df_vendas[
    df_vendas["Country"] == "United Kingdom"
]["InvoiceNo"].nunique()

participacao = (pedidos_reino_unido / total_pedidos) * 100

print("Total de pedidos:", total_pedidos)
print("Pedidos no Reino Unido:", pedidos_reino_unido)
print("Participação do Reino Unido:", round(participacao, 2), "%")
```

### Participação dos pedidos por país

A base de vendas apresentou 19.959 pedidos distintos, dos quais 18.018 pertencem ao Reino Unido.

Isso representa aproximadamente 90,28% do total de pedidos, demonstrando uma forte concentração das operações nesse país.

Os demais países representam, juntos, aproximadamente 9,72% dos pedidos.

```python
valor_medio_pedido = (
    df_vendas.groupby("InvoiceNo")["TotalPrice"]
    .sum()
    .mean()
)

print("Valor médio por pedido: £", round(valor_medio_pedido, 2))
```

### Valor médio por pedido

Foi calculado o valor médio dos pedidos utilizando o agrupamento das transações pelo número da nota fiscal.

Primeiramente, os valores dos produtos foram somados individualmente por pedido. Em seguida, foi calculada a média desses totais.

O valor médio encontrado foi de £ 532,64 por pedido.

```python
pedidos_mensais = (
    df_vendas.assign(
        Mes=df_vendas["InvoiceDate"].dt.to_period("M")
    )
    .groupby("Mes")["InvoiceNo"]
    .nunique()
)

pedidos_mensais
```

```python
resumo_mensal = pd.DataFrame({
    "Faturamento": faturamento_mensal,
    "QuantidadePedidos": pedidos_mensais
})

resumo_mensal["TicketMedio"] = (
    resumo_mensal["Faturamento"] /
    resumo_mensal["QuantidadePedidos"]
)

resumo_mensal.round(2)
```

### Análise dos indicadores mensais

Foi criada uma tabela com três indicadores: faturamento mensal, quantidade de pedidos e ticket médio.

Novembro de 2011 apresentou o maior faturamento, totalizando £ 1.503.866,78, além da maior quantidade de pedidos, com 2.769.

Dezembro de 2011 apresentou o maior ticket médio, de £ 778,74. Entretanto, esse mês possui dados somente até o dia 9, o que limita sua comparação com os demais meses.

A análise permite acompanhar a evolução das vendas e identificar os períodos de maior movimentação comercial.

```python
# Gráficos complementares com os indicadores já calculados.
dados_resumo = resumo_mensal.reset_index()
dados_resumo["Mes"] = dados_resumo["Mes"].astype(str)

# Pergunta: como a quantidade de pedidos varia ao longo dos meses?
fig = px.line(
    dados_resumo,
    x="Mes",
    y="QuantidadePedidos",
    markers=True,
    title="Quantidade de pedidos por mês",
    labels={"Mes": "Mês", "QuantidadePedidos": "Quantidade de pedidos"}
)
fig.show()

# Pergunta: como se distribuem os valores totais dos pedidos?
valores_pedidos = (
    df_vendas.groupby("InvoiceNo")["TotalPrice"]
    .sum()
    .reset_index(name="ValorPedido")
)
fig = px.histogram(
    valores_pedidos,
    x="ValorPedido",
    nbins=40,
    title="Distribuição do valor dos pedidos",
    labels={"ValorPedido": "Valor do pedido (£)"}
)
fig.update_layout(yaxis_title="Quantidade de pedidos")
fig.show()

print("Mediana do valor dos pedidos: £", round(valores_pedidos["ValorPedido"].median(), 2))
print("Média do valor dos pedidos: £", round(valores_pedidos["ValorPedido"].mean(), 2))
print("Todos os pedidos foram preservados, inclusive os de valor elevado.")

# Pergunta: meses com mais pedidos também apresentam maior ticket médio?
fig = px.scatter(
    dados_resumo,
    x="QuantidadePedidos",
    y="TicketMedio",
    hover_name="Mes",
    title="Quantidade de pedidos × ticket médio mensal",
    labels={
        "QuantidadePedidos": "Quantidade de pedidos",
        "TicketMedio": "Ticket médio (£)"
    }
)
fig.show()
```

### Interpretação dos gráficos complementares

O gráfico de linhas permite acompanhar a quantidade de pedidos por mês e comparar os períodos de maior e menor movimento. Como observado na análise mensal, novembro de 2011 teve a maior quantidade de pedidos.

O histograma mostra a frequência dos pedidos em cada faixa de valor. A mediana e a média, calculadas acima, ajudam a verificar se os pedidos de valor elevado influenciam a média. Nenhum pedido foi excluído para construir o gráfico.

Na dispersão, cada ponto representa um mês. A comparação dos indicadores já calculados mostra que o mês com mais pedidos não é o mesmo que apresenta o maior ticket médio. Essa comparação não comprova uma relação de causa e efeito. Dezembro de 2011 possui dados somente até o dia 9 e deve ser interpretado com cuidado.

## Dashboard interativo

A célula abaixo inicia o Streamlit em segundo plano e mostra o link quando o dashboard estiver pronto. Para encerrar o servidor, execute `processo_dashboard.terminate()` em uma célula.

```python
from pathlib import Path
from time import monotonic, sleep
from urllib.error import URLError
from urllib.request import urlopen

from IPython.display import Markdown, display

arquivo_dashboard = Path("dashboard.py").resolve()
host, porta = "127.0.0.1", 8501
url_dashboard = f"http://{host}:{porta}"

if "processo_dashboard" not in globals() or processo_dashboard.poll() is not None:
    processo_dashboard = subprocess.Popen(
        [
            sys.executable, "-m", "streamlit", "run", str(arquivo_dashboard),
            "--server.headless=true",
            f"--server.address={host}",
            f"--server.port={porta}",
            "--browser.gatherUsageStats=false",
        ],
        cwd=arquivo_dashboard.parent,
    )

limite = monotonic() + 30
while monotonic() < limite:
    if processo_dashboard.poll() is not None:
        raise RuntimeError(f"O Streamlit encerrou com o código {processo_dashboard.returncode}.")
    try:
        with urlopen(f"{url_dashboard}/_stcore/health", timeout=1) as resposta:
            if resposta.status == 200:
                break
    except (URLError, TimeoutError):
        sleep(0.25)
else:
    processo_dashboard.terminate()
    processo_dashboard.wait()
    raise TimeoutError("O Streamlit não ficou disponível em 30 segundos.")

display(Markdown(f"[Abrir o dashboard]({url_dashboard})"))
```
